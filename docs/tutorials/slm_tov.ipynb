```python
import numpy as np
import slmemulator as slm
from slmemulator import TOV, SLM, get_paths
import matplotlib.pyplot as plt
```

```python
paths = get_paths()
eos_data_dir = paths['package_eos_data_dir']

# Create a TOV object
tov1 = TOV(eos_data_dir/'apr_eos.table', tidal=True)

# results = tov1.tov_routine()
pvals = np.geomspace(1e-3, 2.0, 30)
Sols_varying_p0 = []
for p0 in pvals:
    results = tov1.tovsolve(p0)
    Sols_varying_p0.append(results)
# print(results)
# plot the results
plt.figure(figsize=(10, 6))
for sol, p0 in zip(Sols_varying_p0, pvals):
    plt.plot(sol[0], sol[2], label=f'p0={p0}')
plt.xlabel('Radius (km)')
plt.ylabel('Mass (M_sun)')
plt.title('TOV Solution')
plt.legend()
# plt.grid()
plt.show()
```
